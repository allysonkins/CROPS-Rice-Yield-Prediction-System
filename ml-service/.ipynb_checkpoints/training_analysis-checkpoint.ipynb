```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    r2_score, mean_absolute_error, mean_squared_error,
    mean_absolute_percentage_error
)

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10, 4)

RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)
print('✅ Imports loaded')
```

```python
df = pd.read_csv('rice_yield_dataset.csv')
df = df.dropna(subset=['yield_tons_ha'])

print(f'Dataset shape: {df.shape}')
print(f'Avg yield: {df["yield_tons_ha"].mean():.2f} t/ha')
df.head()
```

```python
CAT_COLS = ['variety', 'soil_type', 'season', 'seeding_method']
NUM_COLS = ['temperature_avg', 'rainfall_mm', 'humidity_avg',
            'fertilizer_kg_ha', 'historical_yield_tons_ha']

X = df.drop('yield_tons_ha', axis=1)
y = df['yield_tons_ha']

label_encoders = {}
for col in CAT_COLS:
    le = LabelEncoder()
    X[col] = le.fit_transform(X[col])
    label_encoders[col] = le

scaler = StandardScaler()
X[NUM_COLS] = scaler.fit_transform(X[NUM_COLS])

feature_names = NUM_COLS + CAT_COLS
X = X[feature_names]

print(f'Features: {feature_names}')
print(f'X shape: {X.shape}')
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_SEED
)
print(f'Train: {X_train.shape[0]} rows')
print(f'Test:  {X_test.shape[0]} rows')
```

```python
TREES_PER_EPOCH = 10
MAX_EPOCHS = 20

model = RandomForestRegressor(
    n_estimators=TREES_PER_EPOCH,
    max_depth=15,
    min_samples_split=2,
    warm_start=True,
    random_state=RANDOM_SEED,
    n_jobs=-1
)

history = {k: [] for k in ['epoch', 'n_trees', 'train_r2', 'val_r2',
                            'train_mae', 'val_mae', 'train_rmse', 'val_rmse',
                            'train_mape', 'val_mape', 'train_nrmse', 'val_nrmse']}

print(f'{"Epoch":>5} {"Trees":>6} {"Train R²":>10} {"Val R²":>10} {"Train MAE":>11} {"Val MAE":>11}')
print('-' * 70)

for epoch in range(1, MAX_EPOCHS + 1):
    model.n_estimators = epoch * TREES_PER_EPOCH
    model.fit(X_train, y_train)

    y_train_pred = model.predict(X_train)
    y_val_pred = model.predict(X_test)

    tr_r2 = r2_score(y_train, y_train_pred)
    tr_mae = mean_absolute_error(y_train, y_train_pred)
    tr_rmse = np.sqrt(mean_squared_error(y_train, y_train_pred))
    tr_mape = mean_absolute_percentage_error(y_train, y_train_pred) * 100
    tr_nrmse = (tr_rmse / y_train.mean()) * 100

    va_r2 = r2_score(y_test, y_val_pred)
    va_mae = mean_absolute_error(y_test, y_val_pred)
    va_rmse = np.sqrt(mean_squared_error(y_test, y_val_pred))
    va_mape = mean_absolute_percentage_error(y_test, y_val_pred) * 100
    va_nrmse = (va_rmse / y_test.mean()) * 100

    history['epoch'].append(epoch)
    history['n_trees'].append(model.n_estimators)
    history['train_r2'].append(tr_r2); history['val_r2'].append(va_r2)
    history['train_mae'].append(tr_mae); history['val_mae'].append(va_mae)
    history['train_rmse'].append(tr_rmse); history['val_rmse'].append(va_rmse)
    history['train_mape'].append(tr_mape); history['val_mape'].append(va_mape)
    history['train_nrmse'].append(tr_nrmse); history['val_nrmse'].append(va_nrmse)

    print(f'{epoch:>5} {model.n_estimators:>6} {tr_r2:>10.4f} {va_r2:>10.4f} {tr_mae:>11.4f} {va_mae:>11.4f}')

print('\n✅ Training complete')
```

```python
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(history['epoch'], history['train_r2'], marker='o', label='Train R²', color='#0f4c2b', linewidth=2)
ax.plot(history['epoch'], history['val_r2'], marker='s', label='Validation R²', color='#b8860b', linewidth=2)
ax.axhline(y=0.85, color='red', linestyle='--', alpha=0.6, label='Target R² (0.85)')
ax.set_xlabel('Epoch (each = 10 trees)')
ax.set_ylabel('R² Score')
ax.set_title('R² Progression: Train vs Validation')
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('training_r2.png', dpi=100, bbox_inches='tight')
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 4))

axes[0].plot(history['epoch'], history['train_mae'], marker='o', label='Train MAE', color='#0f4c2b')
axes[0].plot(history['epoch'], history['val_mae'], marker='s', label='Val MAE', color='#b8860b')
axes[0].axhline(y=0.4, color='red', linestyle='--', alpha=0.6, label='Target (0.4)')
axes[0].set_xlabel('Epoch'); axes[0].set_ylabel('MAE (t/ha)')
axes[0].set_title('MAE Progression'); axes[0].legend(); axes[0].grid(True, alpha=0.3)

axes[1].plot(history['epoch'], history['train_rmse'], marker='o', label='Train RMSE', color='#0f4c2b')
axes[1].plot(history['epoch'], history['val_rmse'], marker='s', label='Val RMSE', color='#b8860b')
axes[1].set_xlabel('Epoch'); axes[1].set_ylabel('RMSE (t/ha)')
axes[1].set_title('RMSE Progression'); axes[1].legend(); axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('training_errors.png', dpi=100, bbox_inches='tight')
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 4))

axes[0].plot(history['epoch'], history['train_mape'], marker='o', label='Train MAPE', color='#0f4c2b')
axes[0].plot(history['epoch'], history['val_mape'], marker='s', label='Val MAPE', color='#b8860b')
axes[0].axhline(y=8, color='red', linestyle='--', alpha=0.6, label='Target (8%)')
axes[0].set_xlabel('Epoch'); axes[0].set_ylabel('MAPE (%)')
axes[0].set_title('MAPE Progression'); axes[0].legend(); axes[0].grid(True, alpha=0.3)

axes[1].plot(history['epoch'], history['train_nrmse'], marker='o', label='Train NRMSE', color='#0f4c2b')
axes[1].plot(history['epoch'], history['val_nrmse'], marker='s', label='Val NRMSE', color='#b8860b')
axes[1].axhline(y=10, color='red', linestyle='--', alpha=0.6, label='Target (10%)')
axes[1].set_xlabel('Epoch'); axes[1].set_ylabel('NRMSE (%)')
axes[1].set_title('NRMSE Progression'); axes[1].legend(); axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('training_pct_errors.png', dpi=100, bbox_inches='tight')
plt.show()
```

```python
y_pred = model.predict(X_test)

r2 = r2_score(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
mape = mean_absolute_percentage_error(y_test, y_pred) * 100
nrmse = (rmse / y_test.mean()) * 100

print('=' * 55)
print('FINAL REGRESSION EVALUATION (Test Set)')
print('=' * 55)
print(f'  R²     = {r2:.4f}          (target: > 0.85)  {"✅" if r2 > 0.85 else "❌"}')
print(f'  MAE    = {mae:.4f} t/ha    (target: < 0.40)  {"✅" if mae < 0.4 else "❌"}')
print(f'  RMSE   = {rmse:.4f} t/ha')
print(f'  MAPE   = {mape:.2f}%        (target: < 8%)    {"✅" if mape < 8 else "❌"}')
print(f'  NRMSE  = {nrmse:.2f}%        (target: < 10%)   {"✅" if nrmse < 10 else "❌"}')
print('=' * 55)
print(f'  Trees: {model.n_estimators}')
```

```python
fig, ax = plt.subplots(figsize=(7, 6))
ax.scatter(y_test, y_pred, alpha=0.5, color='#0f4c2b', edgecolor='white', s=40)
lims = [min(y_test.min(), y_pred.min()), max(y_test.max(), y_pred.max())]
ax.plot(lims, lims, 'r--', linewidth=2, label='Perfect prediction')
ax.set_xlabel('Actual Yield (t/ha)')
ax.set_ylabel('Predicted Yield (t/ha)')
ax.set_title(f'Actual vs Predicted (R² = {r2:.3f})')
ax.legend(); ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('actual_vs_predicted.png', dpi=100, bbox_inches='tight')
plt.show()
```

```python
importances = model.feature_importances_
indices = np.argsort(importances)[::-1]

fig, ax = plt.subplots(figsize=(9, 5))
ax.bar(range(len(feature_names)), importances[indices], color='#0f4c2b', alpha=0.85)
ax.set_xticks(range(len(feature_names)))
ax.set_xticklabels([feature_names[i] for i in indices], rotation=45, ha='right')
ax.set_ylabel('Importance')
ax.set_title('Feature Importance — Random Forest')
ax.grid(True, alpha=0.3, axis='y')
plt.tight_layout()
plt.savefig('feature_importance.png', dpi=100, bbox_inches='tight')
plt.show()

print('\nFeature importance ranking:')
for i in indices:
    print(f'  {feature_names[i]:<30} {importances[i]:.4f}')
```

```python
joblib.dump(model, 'model_rf.pkl')
joblib.dump(scaler, 'scaler.pkl')
joblib.dump(label_encoders, 'label_encoders.pkl')
joblib.dump(feature_names, 'feature_names.pkl')

print('✅ Model files saved')
```
