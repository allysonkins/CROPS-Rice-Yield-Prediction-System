```python
# =====================================================================
# RICE YIELD REGRESSION — TUNED XGBoost vs baseline ensembles
# =====================================================================
import os, warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

from sklearn.model_selection import (train_test_split, KFold,
                                     cross_val_score, RandomizedSearchCV)
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (mean_squared_error, mean_absolute_error,
                             r2_score, mean_absolute_percentage_error)
from sklearn.ensemble import (RandomForestRegressor,
                              GradientBoostingRegressor,
                              ExtraTreesRegressor)
from xgboost import XGBRegressor

# ---------------- config ----------------
CSV_PATH     = "rice_yield_dataset.csv"
OUT          = "figures"
TARGET       = "yield_tons_ha"
RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)
sns.set_theme(style="whitegrid", context="talk")
plt.rcParams.update({"figure.dpi": 120, "savefig.dpi": 300,
                     "font.size": 11, "axes.titlesize": 13,
                     "axes.labelsize": 11, "axes.titleweight": "bold"})
os.makedirs(OUT, exist_ok=True)

# ================= 1. LOAD =================
df = pd.read_csv(CSV_PATH)
print("Shape:", df.shape)
display(df.head())
display(df[TARGET].describe().round(3))

# ================= 2. EDA =================
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
sns.histplot(df[TARGET], kde=True, bins=40, color="#2E86AB", ax=axes[0])
axes[0].set_title("Distribution of Rice Yield"); axes[0].set_xlabel("Yield (t/ha)")
sns.boxplot(x=df[TARGET], color="#F18F01", ax=axes[1])
axes[1].set_title("Boxplot of Rice Yield"); axes[1].set_xlabel("Yield (t/ha)")
plt.tight_layout(); plt.savefig(f"{OUT}/fig01_target_distribution.png", bbox_inches="tight"); plt.show()

num_cols = ["variety_maturity_days","variety_max_yield","variety_avg_yield",
            "land_area_ha","fertilizer_kg_ha","historical_yield_tons_ha",
            "temperature_avg","rainfall_mm","humidity_avg", TARGET]
fig, ax = plt.subplots(figsize=(9, 7.5))
sns.heatmap(df[num_cols].corr(), annot=True, fmt=".2f", cmap="RdBu_r",
            center=0, square=True, cbar_kws={"shrink": .8},
            annot_kws={"size": 9}, ax=ax)
ax.set_title("Correlation Matrix of Numeric Features")
plt.tight_layout(); plt.savefig(f"{OUT}/fig02_correlation_heatmap.png", bbox_inches="tight"); plt.show()

# ================= 3. SPLIT & SCALE =================
FEATURES = [c for c in df.columns if c != TARGET]
X = df[FEATURES].copy(); y = df[TARGET].copy()
NUM_SCALED = ["variety_maturity_days","variety_max_yield","land_area_ha",
              "fertilizer_kg_ha","historical_yield_tons_ha",
              "temperature_avg","rainfall_mm","humidity_avg"]
NUM_SCALED = [c for c in NUM_SCALED if c in X.columns]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=RANDOM_STATE)
scaler = StandardScaler().fit(X_train[NUM_SCALED])
X_train[NUM_SCALED] = scaler.transform(X_train[NUM_SCALED])
X_test[NUM_SCALED]  = scaler.transform(X_test[NUM_SCALED])
print("Train:", X_train.shape, " Test:", X_test.shape)

# ================= 4a. TUNE XGBoost =================
print("\n>>> Tuning XGBoost via RandomizedSearchCV (40 iters, 3-fold CV)...")

xgb_param_dist = {
    "n_estimators":     [400, 600, 800, 1000, 1500, 2000],
    "learning_rate":    [0.01, 0.02, 0.03, 0.05, 0.08],
    "max_depth":        [4, 5, 6, 7, 8, 9, 10],
    "min_child_weight": [1, 2, 3, 5, 8],
    "subsample":        [0.6, 0.7, 0.8, 0.9, 1.0],
    "colsample_bytree": [0.6, 0.7, 0.8, 0.9, 1.0],
    "reg_alpha":        [0.0, 0.001, 0.01, 0.1],
    "reg_lambda":       [0.0, 0.1, 0.5, 1.0, 2.0],
    "gamma":            [0.0, 0.01, 0.1, 0.3],
}

xgb_search = RandomizedSearchCV(
    XGBRegressor(random_state=RANDOM_STATE, n_jobs=-1, tree_method="hist"),
    param_distributions=xgb_param_dist,
    n_iter=40, scoring="neg_root_mean_squared_error",
    cv=3, random_state=RANDOM_STATE, n_jobs=-1, verbose=0
)
xgb_search.fit(X_train, y_train)
xgb_best = xgb_search.best_estimator_

print("Best XGBoost params:", xgb_search.best_params_)
print(f"Best CV RMSE: {-xgb_search.best_score_:.4f}")

# ================= 4b. FINAL MODELS =================
models = {
    "XGBoost":           xgb_best,  # <- tuned
    "Random Forest":     RandomForestRegressor(n_estimators=500,
                                               random_state=RANDOM_STATE, n_jobs=-1),
    "Gradient Boosting": GradientBoostingRegressor(n_estimators=300,
                                                   learning_rate=0.05, max_depth=3,
                                                   random_state=RANDOM_STATE),
    "Extra Trees":       ExtraTreesRegressor(n_estimators=500,
                                             random_state=RANDOM_STATE, n_jobs=-1),
}

def evaluate(name, model, Xtr, ytr, Xte, yte):
    model.fit(Xtr, ytr)
    pred = model.predict(Xte)
    return {
        "Model": name,
        "RMSE": np.sqrt(mean_squared_error(yte, pred)),
        "MAE":  mean_absolute_error(yte, pred),
        "MAPE (%)": mean_absolute_percentage_error(yte, pred) * 100,
        "R²":   r2_score(yte, pred),
        "_y_pred": pred, "_model": model,
    }

results = [evaluate(n, m, X_train, y_train, X_test, y_test) for n, m in models.items()]
results_df = (pd.DataFrame(results).drop(columns=["_y_pred","_model"])
              .sort_values("RMSE").reset_index(drop=True))
print("\n=== Test-set performance ===")
display(results_df.round(4))
display(results_df.style
        .format({"RMSE":"{:.4f}","MAE":"{:.4f}","MAPE (%)":"{:.2f}","R²":"{:.4f}"})
        .background_gradient(cmap="Blues_r", subset=["RMSE","MAE","MAPE (%)"])
        .background_gradient(cmap="Greens", subset=["R²"]))

# ================= 5. METRIC BAR CHARTS =================
fig, axes = plt.subplots(2, 2, figsize=(13, 9))
metrics = ["RMSE","MAE","MAPE (%)","R²"]
colors  = ["#E63946","#F4A261","#2A9D8F","#264653"]
for ax, metric, c in zip(axes.ravel(), metrics, colors):
    order = results_df.sort_values(metric, ascending=(metric != "R²"))
    sns.barplot(data=order, x="Model", y=metric, ax=ax,
                color=c, edgecolor="black", linewidth=0.6)
    ax.set_title(f"{metric} by Model"); ax.set_xlabel("")
    ax.tick_params(axis="x", rotation=20)
    for i, v in enumerate(order[metric].values):
        ax.text(i, v, f"{v:.3f}", ha="center", va="bottom",
                fontsize=9, fontweight="bold")
plt.suptitle("Predictive Performance Comparison of Ensemble Regressors",
             fontsize=15, fontweight="bold", y=1.02)
plt.tight_layout(); plt.savefig(f"{OUT}/fig03_metric_comparison.png", bbox_inches="tight"); plt.show()

# ================= 6. ACTUAL VS PREDICTED =================
fig, axes = plt.subplots(2, 2, figsize=(12, 11))
lo = float(min(y_test.min(), min(r["_y_pred"].min() for r in results)))
hi = float(max(y_test.max(), max(r["_y_pred"].max() for r in results)))
for ax, r in zip(axes.ravel(), results):
    ax.scatter(y_test, r["_y_pred"], s=12, alpha=0.55, color="#2E86AB")
    ax.plot([lo, hi], [lo, hi], "r--", lw=1.6, label="Ideal (y = x)")
    ax.set_xlim(lo, hi); ax.set_ylim(lo, hi)
    ax.set_title(f"{r['Model']}  |  R² = {r['R²']:.3f}")
    ax.set_xlabel("Actual yield (t/ha)"); ax.set_ylabel("Predicted (t/ha)")
    ax.legend(loc="upper left", fontsize=9)
plt.suptitle("Actual vs Predicted Yield — Test Set", fontsize=15,
             fontweight="bold", y=1.0)
plt.tight_layout(); plt.savefig(f"{OUT}/fig04_actual_vs_predicted.png", bbox_inches="tight"); plt.show()

# ================= 7. RESIDUALS =================
fig, axes = plt.subplots(2, 2, figsize=(13, 9))
for ax, r in zip(axes.ravel(), results):
    resid = y_test.values - r["_y_pred"]
    sns.histplot(resid, kde=True, bins=40, color="#F18F01", ax=ax)
    ax.axvline(0, color="red", ls="--", lw=1.4)
    ax.set_title(f"{r['Model']} residuals  (μ={resid.mean():.3f}, σ={resid.std():.3f})")
    ax.set_xlabel("Residual"); ax.set_ylabel("Count")
plt.suptitle("Residual Distributions", fontsize=15, fontweight="bold", y=1.0)
plt.tight_layout(); plt.savefig(f"{OUT}/fig05_residuals.png", bbox_inches="tight"); plt.show()

# ================= 8. FEATURE IMPORTANCE =================
k = 15
fig, axes = plt.subplots(2, 2, figsize=(14, 11))
for ax, r in zip(axes.ravel(), results):
    m = r["_model"]
    if not hasattr(m, "feature_importances_"): ax.axis("off"); continue
    imp = (pd.Series(m.feature_importances_, index=FEATURES)
           .sort_values(ascending=False).head(k)[::-1])
    ax.barh(imp.index, imp.values, color="#264653")
    ax.set_title(f"{r['Model']} — Top {k}"); ax.set_xlabel("Importance")
plt.suptitle("Feature Importance Across Models", fontsize=15,
             fontweight="bold", y=1.0)
plt.tight_layout(); plt.savefig(f"{OUT}/fig06_feature_importance.png", bbox_inches="tight"); plt.show()

# ================= 9. 5-FOLD CV =================
cv = KFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
cv_rows = []
for name, model in models.items():
    scores = cross_val_score(model, X_train, y_train,
                             scoring="neg_root_mean_squared_error", cv=cv, n_jobs=-1)
    cv_rows.append({"Model": name, "CV RMSE mean": -scores.mean(),
                    "CV RMSE std": scores.std()})
cv_df = pd.DataFrame(cv_rows).sort_values("CV RMSE mean").reset_index(drop=True)
display(cv_df.round(4))
fig, ax = plt.subplots(figsize=(9, 4.8))
sns.barplot(data=cv_df, x="Model", y="CV RMSE mean",
            color="#457B9D", edgecolor="black", ax=ax)
ax.errorbar(x=range(len(cv_df)), y=cv_df["CV RMSE mean"],
            yerr=cv_df["CV RMSE std"], fmt="none",
            ecolor="black", capsize=6, lw=1.6)
ax.set_title("5-Fold CV RMSE"); ax.set_ylabel("RMSE (t/ha)")
ax.tick_params(axis="x", rotation=20)
for i, v in enumerate(cv_df["CV RMSE mean"]):
    ax.text(i, v, f"{v:.3f}", ha="center", va="bottom", fontsize=9, fontweight="bold")
plt.tight_layout(); plt.savefig(f"{OUT}/fig07_cv_rmse.png", bbox_inches="tight"); plt.show()

# ================= 10. RANKING =================
comparison = results_df.set_index("Model")[["RMSE","MAE","MAPE (%)","R²"]].round(4)
display(comparison)

print("\n🏆 BEST PER METRIC")
print(f"  RMSE : {comparison['RMSE'].idxmin()} ({comparison['RMSE'].min():.4f})")
print(f"  MAE  : {comparison['MAE'].idxmin()} ({comparison['MAE'].min():.4f})")
print(f"  MAPE : {comparison['MAPE (%)'].idxmin()} ({comparison['MAPE (%)'].min():.2f}%)")
print(f"  R²   : {comparison['R²'].idxmax()} ({comparison['R²'].max():.4f})")

norm = comparison.copy()
norm["RMSE"]     = (norm["RMSE"]     - norm["RMSE"].min())     / (norm["RMSE"].max()     - norm["RMSE"].min()     + 1e-9)
norm["MAE"]      = (norm["MAE"]      - norm["MAE"].min())      / (norm["MAE"].max()      - norm["MAE"].min()      + 1e-9)
norm["MAPE (%)"] = (norm["MAPE (%)"] - norm["MAPE (%)"].min()) / (norm["MAPE (%)"].max() - norm["MAPE (%)"].min() + 1e-9)
norm["R²"]       = (norm["R²"].max() - norm["R²"])             / (norm["R²"].max()       - norm["R²"].min()       + 1e-9)
norm["Overall"]  = norm.mean(axis=1)
print("\n📊 OVERALL RANKING")
for rank, (model, score) in enumerate(norm["Overall"].sort_values().items(), 1):
    mark = "🥇" if rank == 1 else "🥈" if rank == 2 else "🥉" if rank == 3 else "  "
    print(f"  {mark} #{rank}  {model:<20} score = {score:.4f}")

# Grouped bar
fig, ax = plt.subplots(figsize=(13, 5.5))
mets = ["RMSE","MAE","MAPE (%)","R²"]
x = np.arange(len(mets)); w = 0.20
model_names = list(comparison.index)
palette = ["#E63946","#F4A261","#2A9D8F","#264653"]
pdf = comparison[mets].copy()
for c in mets:
    lo_, hi_ = pdf[c].min(), pdf[c].max()
    pdf[c] = (pdf[c] - lo_) / (hi_ - lo_ + 1e-9)
pdf["R²"] = 1 - pdf["R²"]
for i, (model, color) in enumerate(zip(model_names, palette)):
    bars = ax.bar(x + (i - 1.5) * w, pdf.loc[model], w,
                  label=model, color=color, edgecolor="black", linewidth=0.6)
    for b, raw in zip(bars, comparison.loc[model]):
        ax.text(b.get_x()+b.get_width()/2, b.get_height()+0.015,
                f"{raw:.3f}", ha="center", fontsize=7.5)
ax.set_xticks(x); ax.set_xticklabels(mets)
ax.set_ylabel("Normalised score (lower = better)")
ax.set_title("Four-Way Model Comparison Across All Metrics",
             fontweight="bold")
ax.legend(ncol=4, loc="upper center", bbox_to_anchor=(0.5, -0.08), frameon=False)
ax.set_ylim(0, 1.15)
plt.tight_layout(); plt.savefig(f"{OUT}/fig08_model_comparison_grouped.png",
                                bbox_inches="tight"); plt.show()

# ================= 11. SAVE =================
results_df.to_csv(f"{OUT}/model_comparison.csv", index=False)
cv_df.to_csv(f"{OUT}/cv_results.csv", index=False)
comparison.to_csv(f"{OUT}/model_comparison_detailed.csv")
pd.DataFrame([xgb_search.best_params_]).to_csv(f"{OUT}/xgb_best_params.csv", index=False)

print("\nSaved artifacts to ./" + OUT + "/")
for f in sorted(os.listdir(OUT)): print("  -", f)

best = results_df.iloc[0]
print(f"\n🏆 Best: {best['Model']}  |  RMSE={best['RMSE']:.4f}  "
      f"MAE={best['MAE']:.4f}  MAPE={best['MAPE (%)']:.2f}%  R²={best['R²']:.4f}")
```
